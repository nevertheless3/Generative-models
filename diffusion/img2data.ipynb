```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('mevtheneural.webp', cv2.IMREAD_GRAYSCALE)

h, w = img.shape
new_w = 300
new_h = int(h * (new_w / w))
img_resized = cv2.resize(img, (new_w, new_h))

_, thresh = cv2.threshold(img_resized, 80 , 255, cv2.THRESH_BINARY)

y_coords, x_coords = np.where(thresh > 0)

y_coords = -y_coords

X = np.column_stack((x_coords, y_coords))

np.random.seed(314159)
idx = np.random.choice(len(X), size=min(10000, len(X)), replace=False)
X_sampled = X[idx]

plt.figure(figsize=(6 , 6))
plt.scatter(
    X_sampled[:, 0], X_sampled[:, 1], s=1.5
)


```
