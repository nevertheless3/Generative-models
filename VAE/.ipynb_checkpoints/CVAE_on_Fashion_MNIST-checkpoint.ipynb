```python
import torch
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt

transform = transforms.Compose([
    transforms.ToTensor()
])

import torch
import torchvision
import torchvision.transforms as transforms
import torch.nn.functional as F

transform = transforms.Compose([
    transforms.ToTensor()
])

target_transform = transforms.Lambda(
    lambda y: F.one_hot(torch.tensor(y), num_classes=10).float()
)

trainset = torchvision.datasets.FashionMNIST(
    root='./data', 
    train=True, 
    download=True, 
    transform=transform,
    target_transform=target_transform
)

testset = torchvision.datasets.FashionMNIST(
    root='./data', 
    train=False, 
    download=True, 
    transform=transform,
    target_transform=target_transform
)

train_loader = torch.utils.data.DataLoader(
    trainset, 
    batch_size=64, 
    shuffle=True
)

test_loader = torch.utils.data.DataLoader(
    testset, 
    batch_size=64, 
    shuffle=False
)

classes = [
    'T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot'
]

import matplotlib.pyplot as plt

dataiter = iter(train_loader)
images, labels = next(dataiter)

fig, axes = plt.subplots(2, 8, figsize=(12, 4))
for idx, ax in enumerate(axes.flat):
    ax.imshow(images[idx].squeeze(), cmap='gray')
    ax.axis('off')
    # ax.set_title(classes[labels[idx]], fontsize=9)

plt.suptitle("Fashion-MNIST Samples", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
import torch.nn as nn

class VAE(nn.Module):
    def __init__(self):

        super().__init__()
        
        self.encoder = nn.Sequential(
            nn.Linear(784 + 10 , 400),
            nn.ReLU()
        )

        self.mu = nn.Linear(400,2)
        self.logvar = nn.Linear(400,2)

        self.decoder = nn.Sequential(
            nn.Linear(12 , 400),
            nn.ReLU(),
            nn.Linear(400 , 784 ),
        )

    def forward(self , x , y):
        inp = torch.concat([x , y] , dim = 1)
        encoder = self.encoder(inp)

        mu = self.mu(encoder)
        logvar = self.logvar(encoder)

        eps = torch.randn_like(logvar)

        std = torch.exp(0.5 * logvar)
        
        z = mu + eps * std
    
        latent = torch.concat([z , labels ] , dim = 1)
        recon = self.decoder(latent)

        return mu , logvar , recon 
    

model = VAE()
```

```python
import torch.nn.functional as F

def loss_fn(mu , logvar , x , x_recon):

    recon_loss = F.binary_cross_entropy_with_logits(input=x_recon , target=x , reduction='sum')
    
    kl_loss = -0.5 * torch.sum((logvar + 1 - mu ** 2 - torch.exp(logvar)))

    recon_loss /= batch_size 
    kl_loss /= batch_size

    return (recon_loss + kl_loss) / x.size(0)
    
```

```python
learning_rate = 0.01

optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)
```

```python
batch_size  = 64

def train_loop(dataloader, model, loss_fn, optimizer):
    size = len(dataloader.dataset)
    model.train()
    for batch, (X, y) in enumerate(dataloader):
        X = X.reshape(-1 , 784)
        
        mu , logvar , recon = model(X , y)
        loss = loss_fn(mu , logvar , X , recon)


        loss.backward()
        optimizer.step()
        optimizer.zero_grad()

        if batch % 100 == 0:
            loss, current = loss.item(), batch * batch_size + len(X)
            print(f"loss: {loss:>7f}  [{current:>5d}/{size:>5d}]")


def test_loop(dataloader, model, loss_fn):
    model.eval()
    size = len(dataloader.dataset)
    num_batches = len(dataloader)
    test_loss = 0

    with torch.no_grad():
        for X, y in dataloader:
            X = X.reshape(-1 , 784)
            mu , logvar , recon = model(X , y)
            test_loss += loss_fn(mu , logvar , X , recon)
            
    test_loss /= num_batches
    print(f"Avg loss: {test_loss:>8f} \n")
```

```python
for t in range(5):
    print(f"Epoch {t+1}\n-------------------------------")
    train_loop(train_loader, model, loss_fn, optimizer)
    # test_loop(test_loader, model, loss_fn)
print("Done!")
```
